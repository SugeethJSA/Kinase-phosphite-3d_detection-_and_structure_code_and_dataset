```python
import pandas as pd

# Load the Excel file
file_path = r"C:\Users\Tim\Downloads\Modified_TOP_20_FOR_STRUCTURE.xlsx"  # Update the path if necessary
df = pd.read_excel(file_path)

# Preview the first few rows
print(df.head())
```

```python
from Bio import SeqIO

fasta_file = "sequences.fasta"

with open(fasta_file, "w") as f:
    for index, row in df.iterrows():
        seq_id = f">seq{index+1}\n"
        sequence = row['FASTA'] + "\n"  # Ensure correct column name
        f.write(seq_id + sequence)
```

```python
from Bio.Align.Applications import ClustalOmegaCommandline

# Run Clustal Omega (ensure it's installed)
aligned_fasta = "aligned_sequences.fasta"
clustalomega_cline = ClustalOmegaCommandline(infile=fasta_file, outfile=aligned_fasta, verbose=True, auto=True)
clustalomega_cline()
```
